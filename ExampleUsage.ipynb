# Example Usage

```python
import sys
sys.path.append(".")
from Strain_classifier import standardise_strains

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("ExampleDataset.csv")
data.head()
```

It by default, the function expects to normalise values from the assay_strain (obtained from SQL queries). However you can manually enter another 'strain_name' column if you have downloaded your data directly from chembl. 

```python
strains_standardised = standardise_strains(data, 
                                  species="Klebsiella pneumoniae")
```

The function returns your dataframe with a new `func_assigned_strain` column which can then be used for subsequent analysis

```python
plt.figure(figsize=(7,4.5),facecolor="white")
sns.set_style("darkgrid")

sns.barplot(strains_standardised["func_assigned_strain"].value_counts()[1:6])

plt.title("Top strains in the dataset", fontsize=18, fontweight="bold")
plt.show()
```
